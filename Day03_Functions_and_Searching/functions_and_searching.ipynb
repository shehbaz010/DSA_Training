```python
# Reverse the array

arr = [2, 4, 6, 8, 10]

i = 0
j = len(arr) -1

while i < j:
  arr[i], arr[j] = arr[j], arr[i]
  i += 1
  j -= 1

print(arr)
```

```python
def binary_search(arr, s):

  low = 0
  high = len(arr) - 1

  while low <= high:
    mid = (low + high) // 2
    if arr[mid] == s:
      print("Element found at = ",mid)
      return True
      break
    elif arr[mid] < s:
      low = mid + 1
    else:
      high = mid - 1

arr = []

n = int(input("Enter the size of the array: "))

for i in range(n):
    arr.append(int(input(f"Enter ele : ")))

arr.sort()

s = int(input("Enter ele you want to search: "))

print(binary_search(arr, s))
```
