```python
# Merge sort in different size list # with sorted array


class MergeSort:
  def mergesort(self, arr1, arr2):
    i = 0
    j = 0
    k = 0

    while i < len(arr1) and j < len(arr2):
      if arr1[i] < arr2[j]:
        arr3.append(arr1[i])
        i = i + 1
        k = k + 1
      else:
        arr3.append(arr2[j])
        j = j + 1
        k = k + 1

    while len(arr1) > i:
      arr3.append(arr1[i])
      i = i + 1
      k = k + 1

    while len(arr2) > j:
      arr3.append(arr2[j])
      j = j + 1

    return arr3

if __name__ == '__main__':

  arr3 = []
  obj = MergeSort()

  arr1 = [1, 3, 5, 7, 9]
  arr2 = [2, 4, 6, 8]

  ans = obj.mergesort(arr1, arr2)
  print(ans)
```

```python
# Merge sort in unsorted array

class MergeSort:
    def mergesort(self, arr):

        if len(arr) <= 1:
            return arr

        mid = len(arr) // 2

        arr1 = arr[:mid]
        arr2 = arr[mid:]

        self.mergesort(arr1)
        self.mergesort(arr2)

        i = 0
        j = 0
        k = 0

        while i < len(arr1) and j < len(arr2):
            if arr1[i] > arr2[j]:
                arr[k] = arr1[i]
                i += 1
            else:
                arr[k] = arr2[j]
                j += 1
            k += 1

        while i < len(arr1):
            arr[k] = arr1[i]
            i += 1
            k += 1

        while j < len(arr2):
            arr[k] = arr2[j]
            j += 1
            k += 1

        return arr

if __name__ == '__main__':
    obj = MergeSort()

    arr = [111, 212, 32, 3, 98, 45, 5, 1, 7, 11, 9]

    ans = obj.mergesort(arr)
    print("Sorted Array:", ans)
```
