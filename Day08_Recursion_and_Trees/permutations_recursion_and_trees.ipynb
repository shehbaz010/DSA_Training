```python
# Tree

class BST:

    def __init__(self, key=None):
        self.data = key
        self.left = None
        self.right = None


    def insert(self, key):

        if self.data is None:
            self.data = key
            return

        if self.data == key:
            return

        if key < self.data:
            if self.left:
                self.left.insert(key)
            else:
                self.left = BST(key)

        else:
            if self.right:
                self.right.insert(key)
            else:
                self.right = BST(key)




    def Preorder(self):

        print(self.data, end=" -> ")

        if self.left:
            self.left.Preorder()

        if self.right:
            self.right.Preorder()




    def Postorder(self):

        if self.left:
            self.left.Postorder()

        if self.right:
            self.right.Postorder()

        print(self.data, end=" -> ")




    def Inorder(self):

        if self.left:
            self.left.Inorder()

        print(self.data, end=" -> ")

        if self.right:
            self.right.Inorder()




    def search(self, key):

        if self.data is None:
            print("Tree is empty.")
            return False

        if self.data == key:
            print(key, "key found in the tree.")
            return True

        if key < self.data:
            if self.left:
                return self.left.search(key)
            else:
                print(key, "Key not found in the tree.")
                return False

        else:
            if self.right:
                return self.right.search(key)
            else:
                print(key, "Key not found in the tree.")
                return False


if __name__ == '__main__':

    root = BST()

    while True:

        print("\n1. Insert")
        print("2. Preorder")
        print("3. Postorder")
        print("4. Inorder")
        print("5. Search")
        print("6. Exit")

        n = int(input("Enter your choice = "))

        if n == 1:

            arr = [36, 26, 46, 21, 31, 11, 24, 41, 56, 51, 66]

            for i in arr:
                root.insert(i)

            print("Elements inserted:", arr)

        elif n == 2:

            print("Preorder Traversal:")
            root.Preorder()
            print()

        elif n == 3:

            print("Postorder Traversal:")
            root.Postorder()
            print()

        elif n == 4:

            print("Inorder Traversal:")
            root.Inorder()
            print()

        elif n == 5:
            search_key = int(input("Enter the key to search: "))
            root.search(search_key)

        else:
            break
```

```python
# Recursion
# Factorial using recursion

def fact(n=5):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fact(n - 1)

print(fact())
```

```python
# Multiply using recursion

def multiply(x, y):
    if y == 0 or x == 0:
        return 0
    elif y == 1:
        return x
    elif x == 1:
        return y
    else:
        return x + multiply(x, y - 1)

print(multiply(5, 3))
```

```python
# power using recursion

def power(x, y):
    if x == 0:
        return 0
    elif y == 0:
      return 1
    elif y == 1:
        return x
    elif x == 1:
        return 1
    else:
        return x * power(x, y - 1)

print(power(5, 3))
```

```python
# find sum of n using recursion

def sum(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return n + sum(n - 1)
print(sum(4))
```

```python
# find permutation of 1st list and choose the smallest among all combination
# grater than 2nd no and closest to 2nd no

from itertools import permutations

a = [123]
b = [200]

num = str(a[0])
target = b[0]

nums = set()

for i in permutations(num):
    nums.add(int("".join(i)))

greater_nums = [x for x in nums if x > target]

if greater_nums:
    result = min(greater_nums)
    print(result)
else:
    print("No permutation greater than target")
```
