```python
# Reverse of a number using loop

num = int(input("Enter a number: "))
rev = 0

while num > 0:
    rem = num % 10
    rev = rev * 10 + rem
    num = num // 10

print("Reverse =", rev)
```

```python
# Check no is palindrom or not

num = 121
temp = num
rev = 0

while num > 0:
    rem = num % 10
    rev = rev * 10 + rem
    num = num // 10

if temp == rev:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
#Check no is armstrong or not

num = int(input("Enter a no"))
temp = num
sum = 0

count = 0

while num > 0:
    num = num // 10
    count = count + 1

while num > 0:
    rem = num % 10
    sum = sum + (rem ** 3)
    num = num // 10

if temp == sum:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# Print all Armstrong numbers from 1 to 1000

for num in range(1, 1001):
    temp = num
    sum = 0

    while temp > 0:
        rem = temp % 10
        sum = sum + (rem ** 3)
        temp = temp // 10

    if num == sum:
        print(num)
```

```python
#Peterson no

num = 145
temp = num
sum = 0

while num > 0:
    rem = num % 10

    fact = 1
    i = 1
    while i <= rem:
        fact = fact * i
        i = i + 1

    sum = sum + fact
    num = num // 10

if temp == sum:
    print("Peterson Number")
else:
    print("Not a Peterson Number")
```
