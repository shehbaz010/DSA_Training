```python
#Print in matrix form without bracket
arr = [11, 22, 33, 44, 55, 99, 66, 77]

cols = 3

for i in range(0, len(arr), cols):
    for j in range(cols):
        if i + j < len(arr):
            print(arr[i + j], end="\t")
    print()
```

```python
# Count number of even and odd in list

n = int(input("Enter size: "))
arr = []

for i in range(n):
  arr.append(int(input("Enter value: ")))
print(arr)

even = 0
odd = 0

for i in range(len(arr)):
  if arr[i] % 2 == 0:
    even = even + 1
  else:
    odd = odd + 1

print("Even count: ", even)
print("Odd count: ", odd)
```

```python
# Find no is tech no. or not

n = input("Enter a no: ")
num = int(n)
length = len(n)

if length % 2 == 0:
    half = length // 2
    left = int(n[:half])
    right = int(n[half:])

    sum = left + right
    result = (sum) ** 2

    print("Sum =", sum)

    if result == num:
        print(num, "is a Tech num.")
    else:
        print(num, "is not a Tech num.")
else:
    print("Invalid: Tech nums must have an even num of digits.")
```

```python
# accept mobile no, check no is 10 digit, check no. is valid in India or not

num = input("Enter a no : ")

if len(num) != 10:
    print("no. must be 10 digits")

elif not num.isdigit():
    print("no. must have only digits")

elif num[0] not in ('6', '7', '8', '9'):
    print("no. must start with = 6, 7, 8, or 9.")

else:
    print("Valid Indian Mobile no :", num)
```
